```python
import pandas as pd
import numpy as np

# Veri setini okuma
df = pd.read_csv("data.csv", encoding="ISO-8859-1")

print("Veri setinin boyutu:", df.shape)
print("\nEksik değerler:")
print(df.isnull().sum())

# Tekrarlanan satırları silme
tekrar_sayisi = df.duplicated().sum()
df = df.drop_duplicates()

print("\nSilinen tekrar sayısı:", tekrar_sayisi)

# CustomerID ve Description eksik olan satırları silme
df = df.dropna(subset=["CustomerID", "Description"])

print("Eksik değerler temizlendikten sonra:", df.shape)

# İptal edilen siparişleri silme
iptal_sayisi = df["InvoiceNo"].astype(str).str.startswith("C").sum()

df = df[~df["InvoiceNo"].astype(str).str.startswith("C")]

# Negatif veya 0 olan miktar ve fiyatları temizleme
df = df[(df["Quantity"] > 0) & (df["UnitPrice"] > 0)]

print("İptal edilen işlem sayısı:", iptal_sayisi)
print("Temizlik sonrası veri boyutu:", df.shape)

# Veri tiplerini düzenleme
df["CustomerID"] = df["CustomerID"].astype(int).astype(str)
df["InvoiceDate"] = pd.to_datetime(df["InvoiceDate"])

# Yeni sütunlar oluşturma
df["TotalPrice"] = df["Quantity"] * df["UnitPrice"]
df["YearMonth"] = df["InvoiceDate"].dt.to_period("M").astype(str)
df["Hour"] = df["InvoiceDate"].dt.hour
df["DayOfWeek"] = df["InvoiceDate"].dt.day_name()

print("\nTemizlenmiş veri setinin özeti:")
print(df[["Quantity", "UnitPrice", "TotalPrice"]].describe().round(2))
```

```python
# 1. Genel iş metrikleri

toplam_ciro = df["TotalPrice"].sum()
toplam_siparis = df["InvoiceNo"].nunique()
toplam_musteri = df["CustomerID"].nunique()
toplam_urun = df["StockCode"].nunique()
ortalama_sepet = toplam_ciro / toplam_siparis

print("=== GENEL İŞ METRİKLERİ ===")
print(f"Toplam Ciro: £{toplam_ciro:,.2f}")
print(f"Toplam Sipariş Sayısı: {toplam_siparis:,}")
print(f"Toplam Müşteri Sayısı: {toplam_musteri:,}")
print(f"Toplam Ürün Sayısı: {toplam_urun:,}")
print(f"Ortalama Sepet Tutarı: £{ortalama_sepet:,.2f}")


# 2. En çok satış yapılan ilk 10 ülke

ulke_satis = df.groupby("Country").agg(
    Ciro=("TotalPrice", "sum"),
    Siparis=("InvoiceNo", "nunique"),
    Musteri=("CustomerID", "nunique")
)

ulke_satis = ulke_satis.sort_values(by="Ciro", ascending=False).head(10)

ulke_satis["Ciro Orani (%)"] = (
    ulke_satis["Ciro"] / toplam_ciro * 100
).round(2)

print("\n=== EN ÇOK CİRO GETİREN 10 ÜLKE ===")
print(ulke_satis)


# 3. En çok gelir getiren ilk 10 ürün

urun_satis = df.groupby("Description").agg(
    Toplam_Adet=("Quantity", "sum"),
    Toplam_Ciro=("TotalPrice", "sum")
)

urun_satis = urun_satis.sort_values(
    by="Toplam_Ciro", ascending=False
).head(10)

print("\n=== EN ÇOK SATIŞ GETİREN 10 ÜRÜN ===")
print(urun_satis)


# 4. Aylık satış ve sipariş analizi

aylik_satis = df.groupby("YearMonth").agg(
    Aylik_Ciro=("TotalPrice", "sum"),
    Siparis_Sayisi=("InvoiceNo", "nunique")
)

print("\n=== AYLIK SATIŞ ANALİZİ ===")
print(aylik_satis)


# 5. RFM analizi

son_tarih = df["InvoiceDate"].max() + pd.Timedelta(days=1)

rfm = df.groupby("CustomerID").agg(
    Recency=("InvoiceDate", lambda x: (son_tarih - x.max()).days),
    Frequency=("InvoiceNo", "nunique"),
    Monetary=("TotalPrice", "sum")
)

# RFM puanlarının hesaplanması

rfm["R_Score"] = pd.qcut(
    rfm["Recency"], 5, labels=[5, 4, 3, 2, 1]
).astype(int)

rfm["F_Score"] = pd.qcut(
    rfm["Frequency"].rank(method="first"),
    5,
    labels=[1, 2, 3, 4, 5]
).astype(int)

rfm["M_Score"] = pd.qcut(
    rfm["Monetary"], 5, labels=[1, 2, 3, 4, 5]
).astype(int)

# R ve F puanlarını birleştirme
rfm["RF_Score"] = (
    rfm["R_Score"].astype(str) + rfm["F_Score"].astype(str)
)

# Müşteri gruplarının belirlenmesi

segmentler = {
    r"[1-2][1-2]": "Uyuyan Müşteriler",
    r"[1-2][3-4]": "Risk Altındaki Müşteriler",
    r"[1-2]5": "Kaybedilmemesi Gerekenler",
    r"3[1-2]": "Uykuya Dalanlar",
    r"33": "Dikkat Gerektirenler",
    r"[3-4][4-5]": "Sadık Müşteriler",
    r"41": "Umut Vaat Edenler",
    r"51": "Yeni Müşteriler",
    r"[4-5][2-3]": "Potansiyel Sadıklar",
    r"5[4-5]": "Şampiyonlar"
}

rfm["Segment"] = rfm["RF_Score"].replace(segmentler, regex=True)


# Segmentlerin genel durumunu inceleme

segment_ozet = rfm.groupby("Segment").agg(
    Musteri_Sayisi=("Recency", "count"),
    Ortalama_Recency=("Recency", "mean"),
    Ortalama_Frequency=("Frequency", "mean"),
    Ortalama_Monetary=("Monetary", "mean"),
    Toplam_Ciro=("Monetary", "sum")
)

segment_ozet = segment_ozet.sort_values(
    by="Toplam_Ciro", ascending=False
).round(2)

segment_ozet["Ciro Orani (%)"] = (
    segment_ozet["Toplam_Ciro"] /
    segment_ozet["Toplam_Ciro"].sum() * 100
).round(2)

print("\n=== RFM SEGMENT SONUÇLARI ===")
print(segment_ozet)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.style.use("seaborn-v0_8-whitegrid")


# 1. E-ticaret performans panosu

fig, axes = plt.subplots(2, 2, figsize=(18, 12))

fig.suptitle(
    "E-Ticaret Satış Performansı (2010-2011)",
    fontsize=16
)


# Aylık ciro grafiği

ax1 = axes[0, 0]

ax1.plot(
    aylik_satis.index,
    aylik_satis["Aylik_Ciro"] / 1000,
    marker="o"
)

ax1.set_title("Aylık Ciro Değişimi")
ax1.set_xlabel("Ay")
ax1.set_ylabel("Ciro (Bin £)")
ax1.tick_params(axis="x", rotation=45)


# Ülkelere göre ciro grafiği

ax2 = axes[0, 1]

ulke_veri = df[df["Country"] != "United Kingdom"]

ulke_veri = ulke_veri.groupby("Country")["TotalPrice"].sum()
ulke_veri = ulke_veri.sort_values(ascending=True).tail(10) / 1000

ax2.barh(ulke_veri.index, ulke_veri.values)

ax2.set_title("İngiltere Dışında En Çok Ciro Getiren 10 Ülke")
ax2.set_xlabel("Ciro (Bin £)")


# En çok gelir getiren ürünler

ax3 = axes[1, 0]

urun_grafik = urun_satis.sort_values(
    by="Toplam_Ciro", ascending=True
)

urun_grafik = urun_grafik.tail(10)

ax3.barh(
    urun_grafik.index,
    urun_grafik["Toplam_Ciro"] / 1000
)

ax3.set_title("En Çok Ciro Getiren 10 Ürün")
ax3.set_xlabel("Ciro (Bin £)")


# Gün ve saatlere göre sipariş yoğunluğu

ax4 = axes[1, 1]

gunler = [
    "Monday", "Tuesday", "Wednesday",
    "Thursday", "Friday", "Sunday"
]

gun_isimleri = [
    "Pazartesi", "Salı", "Çarşamba",
    "Perşembe", "Cuma", "Pazar"
]

siparis_veri = df.pivot_table(
    index="DayOfWeek",
    columns="Hour",
    values="InvoiceNo",
    aggfunc="nunique"
)

siparis_veri = siparis_veri.reindex(gunler).fillna(0)
siparis_veri.index = gun_isimleri

sns.heatmap(
    siparis_veri,
    cmap="YlGnBu",
    annot=True,
    fmt=".0f",
    ax=ax4
)

ax4.set_title("Gün ve Saatlere Göre Sipariş Yoğunluğu")
ax4.set_xlabel("Saat")
ax4.set_ylabel("Gün")

plt.tight_layout()

plt.savefig("ecommerce_dashboard.png", dpi=300)
plt.show()


# 2. RFM müşteri segmentasyonu grafiği

plt.figure(figsize=(12, 6))

segment_grafik = segment_ozet.sort_values(
    by="Musteri_Sayisi",
    ascending=True
)

plt.barh(
    segment_grafik.index,
    segment_grafik["Musteri_Sayisi"]
)

plt.title("RFM Müşteri Segmentasyonu")
plt.xlabel("Müşteri Sayısı")
plt.ylabel("Müşteri Segmentleri")

plt.tight_layout()

plt.savefig("rfm_segments.png", dpi=300)
plt.show()
```
